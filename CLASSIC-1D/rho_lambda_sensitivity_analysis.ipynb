```python
# Env: sc2_v0
import xarray as xr
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import proplot as pplt # New plot library (https://proplot.readthedocs.io/en/latest/)
pplt.rc['savefig.dpi'] = 300 # 1200 is too big! #https://proplot.readthedocs.io/en/latest/basics.html#Creating-figures
from scipy import stats
import xskillscore as xs
import calendar
import os
```

```python
def calonne(rho):
    """
    Calculate the Calonne parameter based on snow density.

    Parameters:
    rho (float): Snow density in kg/m^3.

    Returns:
    float: Calonne parameter.
    """
    return 2.5e-6 * rho**2 - 1.23e-4 * rho + 0.024
```

```python
h1 = 0.02
h2 = 0.02

rho1_low = 300
rho2_low = 300

rho1_mid = 250
rho2_mid = 350

rho1_high = 200
rho2_high = 400

rho1_extr = 150
rho2_extr = 450

rho_low = np.average([rho1_low, rho2_low])
rho_mid = np.average([rho1_mid, rho2_mid])
rho_high = np.average([rho1_high, rho2_high])
rho_extr = np.average([rho1_extr, rho2_extr])

print(f"rho_low: {rho_low}, rho_mid: {rho_mid}, rho_high: {rho_high}, rho_extr: {rho_extr}")

lambda_bulk_low = calonne(rho_low)
lambda_bulk_mid = calonne(rho_mid)
lambda_bulk_high = calonne(rho_high)
lambda_bulk_extr = calonne(rho_extr)
print(f"lambda_bulk_low: {lambda_bulk_low}, lambda_bulk_mid: {lambda_bulk_mid}, lambda_bulk_high: {lambda_bulk_high}, lambda_bulk_extr: {lambda_bulk_extr}")

lambda1_low = calonne(rho1_low)
lambda2_low = calonne(rho2_low)
lambda1_mid = calonne(rho1_mid)
lambda2_mid = calonne(rho2_mid)
lambda1_high = calonne(rho1_high)
lambda2_high = calonne(rho2_high)
lambda1_extr = calonne(rho1_extr)
lambda2_extr = calonne(rho2_extr)

lambda_profile_low = (h1 + h2)/(h1 / lambda1_low + h2 / lambda2_low)
lambda_profile_mid = (h1 + h2)/(h1 / lambda1_mid + h2 / lambda2_mid)
lambda_profile_high = (h1 + h2)/(h1 / lambda1_high + h2 / lambda2_high)
lambda_profile_extr = (h1 + h2)/(h1 / lambda1_extr + h2 / lambda2_extr)
print()
print(f"lambda1_low: {lambda1_low}, lambda2_low: {lambda2_low}, lambda_profile_low: {lambda_profile_low}")
print(f"lambda1_mid: {lambda1_mid}, lambda2_mid: {lambda2_mid}, lambda_profile_mid: {lambda_profile_mid}")
print(f"lambda1_high: {lambda1_high}, lambda2_high: {lambda2_high}, lambda_profile_high: {lambda_profile_high}")
print(f"lambda1_extr: {lambda1_extr}, lambda2_extr: {lambda2_extr}, lambda_profile_extr: {lambda_profile_extr}")

r_profile_low = h1 / lambda1_low + h2 / lambda2_low
r_profile_mid = h1 / lambda1_mid + h2 / lambda2_mid
r_profile_high = h1 / lambda1_high + h2 / lambda2_high
r_profile_extr = h1 / lambda1_extr + h2 / lambda2_extr

r_bulk_low = (h1 + h2) / lambda_bulk_low
r_bulk_mid = (h1 + h2) / lambda_bulk_mid
r_bulk_high = (h1 + h2) / lambda_bulk_high
r_bulk_extr = (h1 + h2) / lambda_bulk_extr
print()
print(f"r_profile_low: {r_profile_low}, r_bulk_low: {r_bulk_low}")
print(f"r_profile_mid: {r_profile_mid}, r_bulk_mid: {r_bulk_mid}")
print(f"r_profile_high: {r_profile_high}, r_bulk_high: {r_bulk_high}")
print(f"r_profile_extr: {r_profile_extr}, r_bulk_extr: {r_bulk_extr}")

err_low = (r_profile_low - r_bulk_low) / r_profile_low * 100
err_mid = (r_profile_mid - r_bulk_mid) / r_profile_mid * 100
err_high = (r_profile_high - r_bulk_high) / r_profile_high * 100
err_extr = (r_profile_extr - r_bulk_extr) / r_profile_extr * 100
print()
print(f"err_low: {err_low:.2f}%, err_mid: {err_mid:.2f}%, err_high: {err_high:.2f}%, err_extr: {err_extr:.2f}%")
```
